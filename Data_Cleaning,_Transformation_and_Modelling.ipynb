1.  What is Data Loading in Power BI and why is it considered the first step of analysis?

-   Data Loading in Power BI is the process of importing data from different sources such as Excel, CSV files, databases, or web sources into Power BI.

It is considered the first step of analysis because the data must be brought into Power BI before it can be cleaned, transformed, visualized, and analyzed.

Example:
If we have a Real Estate Dataset in Excel, we first load the Excel file into Power BI. After loading, we can clean the data, create relationships, and build reports and dashboards.

2.  Explain the difference between “Load” and “Transform Data” in Power BI.
-  | Load                                           | Transform Data                                          |
| ---------------------------------------------- | ------------------------------------------------------- |
| Loads the data directly into Power BI.         | Opens Power Query Editor to clean and modify the data.  |
| No major changes are made to the data.         | Data can be filtered, renamed, merged, split, etc.      |
| Used when the data is already clean and ready. | Used when the data needs cleaning or preparation.       |
| Faster and simpler.                            | Provides many data-cleaning and transformation options. |




3.  What is a Fact Table and a Dimension Table? Give examples from the dataset.
Fact Table

-   A Fact Table contains measurable or transactional data that is used for analysis. It usually contains numerical values and keys that connect it to dimension tables.

Example from a Real Estate Dataset:

Property ID
Price
Area
Number of Bedrooms
Number of Bathrooms
Property Type ID
Location ID

For example, Property Price and Area can be analyzed to calculate totals, averages, or comparisons.

Dimension Table

A Dimension Table contains descriptive information that helps explain or categorize the data in the fact table.

Examples:

Property Dimension: Property Type, Bedrooms, Bathrooms
Location Dimension: City, State, Locality
Date Dimension: Date, Month, Year

Simple example:

Fact_Property → Property ID, Location ID, Price, Area

Dim_Location → Location ID, City, State, Locality

The dimension table provides descriptive details about the facts.

 4.  Why is Star Schema preferred over Snowflake Schema in Power BI?

-   A Star Schema is preferred in Power BI because it is simple, faster, and easier to understand and manage.

Advantages of Star Schema:
Simple structure – One fact table is connected directly to multiple dimension tables.
Better performance – Fewer relationships and simpler queries can improve report performance.
Easy to understand – It is easier to create visuals and measures.
Easier DAX calculations – Relationships between fact and dimension tables are straightforward.
Better for reporting – It works well with Power BI's data modelling and filtering.
Example:

Star Schema:

Dim_Location → Fact_Property ← Dim_PropertyType
　　　　　　　　　↓
　　　　　　 Dim_Date

In a Snowflake Schema, dimension tables are further divided into smaller related tables, making the model more complex.

Therefore, Star Schema is generally preferred in Power BI because it provides a clean, efficient, and easy-to-use data model.
